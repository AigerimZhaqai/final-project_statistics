```python
ab = pd.read_excel(file_path, sheet_name='Данные АБ тестов')
ab.head()
```

```python
ab['experiment_num'].unique()
```

```python
for exp in ab['experiment_num'].unique():
    data_exp = ab[ab['experiment_num'] == exp]
    
    group_test = data_exp[data_exp['experiment_group'] == 'test']['revenue']
    group_control = data_exp[data_exp['experiment_group'] == 'control']['revenue']
    
    stat, p = ttest_ind(group_test, group_control, equal_var=False)
    
    print(f'Experiment {exp}: p-value = {p}')
```
